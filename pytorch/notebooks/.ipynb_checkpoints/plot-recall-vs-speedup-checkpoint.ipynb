```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MaxNLocator, FuncFormatter, MultipleLocator
import statistics
```

```python
# RTL plots

models = ["MobileNetV2","GoogLeNet","ShuffleNetX2","ResNet18","ResNet50","InceptionV3","ResNeXt64","ResNeXt32"]
model_alias = ["M1","M2","M3","M4","M5","M6","M7","M8"]

recalls_dac = [98.03, 97.08, 98.00, 94.75, 92.58, 94.74, 90.15, 83.65]
recalls_us  = [99.95, 99.82, 99.97, 99.58, 100.0, 99.45, 99.94, 99.59]

spu_dac = [1.66, 0.97, 1.65, 1.09, 1.89, 1.93, 3.88, 6.33]
spu_us  = [3.38, 2.20, 2.93, 2.20, 3.27, 5.17, 5.79, 7.36]


#color1, color2, color3 = "#322332","#0077b3", "#ff9933"
color1, color2, color3 = "black","#0000ff", "#ff9933"

plt.figure(figsize=(8, 5))

# DAC points

dac_scatter = plt.scatter(spu_dac, recalls_dac, color=color1, label="DAC'16")

plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:.0f}%"))
plt.gca().xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:.0f}×"))

# DAC points

for x, y, label in zip(spu_dac, recalls_dac, model_alias):
    dx, dy = 0, 0
    if label == "M3": dy = -1
    
    plt.text(x + dx, y + dy, label, fontsize=14, color=color1, rotation=0)
        
# Our points
us_scatter = plt.scatter(spu_us, recalls_us, color=color2, label="Proposed", marker="x")

for x, y, label in zip(spu_us, recalls_us, model_alias):
    dx, dy = 0, 0
    #if label == "M1": dx = -0.1;dy = 0.3
    if label == "M2": dx = -0.1; dy = 0.3
    if label == "M3": dx = -0.2; dy = 0.3
    if label == "M4": dy = -1
    if label == "M5": dx = -0.2; dy = -1.2
 
    plt.text(x + dx, y + dy, label, fontsize=14, color=color2, rotation=0)

legend_elements = [
    Line2D([0], [0], marker='o', color='w', label=f"{alias}: {orig}", 
           markerfacecolor='gray', markersize=6)
    for alias, orig in zip(model_alias, models)
]

# --- Legend 1: DAC vs US (scatter markers) ---
legend1 = plt.legend(handles=[us_scatter, dac_scatter],
                     loc="lower left", fontsize=18)
plt.gca().add_artist(legend1)  # keep it when adding second legend

# --- Legend 2: Model mapping ---
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label=f"{alias}: {orig}",
           markerfacecolor='gray', markersize=6)
    for alias, orig in zip(model_alias, models)
]
plt.legend(handles=legend_elements, loc='center right', fontsize=16)

plt.xlabel("Speedup", fontsize=18)
plt.ylabel("Recall (%)", fontsize=18)
plt.xticks(fontsize=18)
plt.yticks(fontsize=18)

plt.grid(True)
plt.tight_layout()
#plt.show()
plt.savefig("reports/plots/date2026/speedup_vs_recall_hw.pdf")
```
