```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MaxNLocator, FuncFormatter, MultipleLocator
import seaborn as sns
import matplotlib.colors as mcolors
import matplotlib.cm as cm
import statistics
```

```python

def plot_clustered_bars(xtics, y1, y2, y_label, y1_std=None, y2_std=None, label1="Series 1", label2="Series 2", save_to="."):
    x = np.arange(len(xtics))  # category positions
    width = 0.35               # width of each bar

    axis_label_size=18
    tick_size=16
    legend_size = 16
    number_on_top_bar_size=12

    #color1, color2 = "rosybrown", "tan" 
    color1, color2 = "#0077b3", "#c6c6c6" 

    fig, ax = plt.subplots(figsize=(8, 4))

    # Bars with error bars if provided
    rects1 = ax.bar(x - width/2, y1, width, label=label1, yerr=y1_std, capsize=5, color=color1, edgecolor="black", hatch="//")
    rects2 = ax.bar(x + width/2, y2, width, label=label2, yerr=y2_std, capsize=5, color=color2, edgecolor="black", hatch='xx')

    # Labels and title
    ax.set_ylabel(y_label, fontsize=axis_label_size) # tamaho to titulo do eixo y. nao dos valores

    ax.set_xticks(x)
    ax.set_xticklabels(xtics)

    #ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f" {val:.0f}×"))
    ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f" {val:.1f}%"))

    # Optional: Add values above bars
    ax.bar_label(rects1, rotation=90, padding=0, fontsize=number_on_top_bar_size, fmt=" %.1f") # recall: padding=6
    ax.bar_label(rects2, rotation=90, padding=0, fontsize=number_on_top_bar_size, fmt=" %.1f") # recall: padding=3

    ax.tick_params(axis='x', labelsize=11, rotation=14)
    ax.tick_params(axis='y', labelsize=tick_size)

    #ax.set_ylim(1, max(max(y1), max(y2)) * 1.12) # speedup
    #ax.set_ylim(0, max(max(y1), max(y2)) * 1.12) # avf
    #ax.set_ylim(80, max(max(y1), max(y2)) * 1.05) # recall

    ax.set_ylim(0, 119.9) # recall
    ax.margins(y=1)   # no vertical margin

    plt.tight_layout(pad=1)
    #ax.legend()
    #ax.legend(loc="lower left", ncol=2, fontsize=legend_size, framealpha=1, facecolor="white") # recall. shift the box a little to the right

    ax.legend(
    loc="lower right",           # anchor legend by its left center
    #bbox_to_anchor=(1, 0.5),     # place it just outside the right edge
    framealpha=1,                # fully opaque
    facecolor="white",
    edgecolor="black",
    fontsize=legend_size
    )
    
    #plt.show()
    plt.savefig(save_to) #, dpi=300, bbox_inches='tight')
```

```python
def plot_clustered_bars3(
    xtics, y1, y2, y3,
    y_label,
    y1_std=None, y2_std=None, y3_std=None,
    label1="Series 1", label2="Series 2", label3="Series 3",
    save_to="."
):
    x = np.arange(len(xtics))  # category positions
    width = 0.25               # narrower width (3 bars per group)

    axis_label_size=18
    tick_size=16
    legend_size = 16
    number_on_top_bar_size=12

    # choose colors
    color1, color2, color3 = "#0077b3", "#c6c6c6", "#ff9933"

    fig, ax = plt.subplots(figsize=(8, 4))

    # Three bar groups per x-tick
    rects1 = ax.bar(x - width, y1, width, label=label1, yerr=y1_std, capsize=5, color=color1, edgecolor="black", hatch="//")
    rects2 = ax.bar(x,         y2, width, label=label2, yerr=y2_std, capsize=5, color=color2, edgecolor="black", hatch="xx")
    rects3 = ax.bar(x + width, y3, width, label=label3, yerr=y3_std, capsize=5, color=color3, edgecolor="black", hatch="\\\\")  # double backslash for escape

    # Label rects1 (centered)
    ax.bar_label(rects1, rotation=90, padding=0, fontsize=number_on_top_bar_size, fmt=" %.2f")

    # Label rects2 (shift slightly left so they don't collide with rects3)
    for bar in rects2:
        height = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2 - bar.get_width() * 0.1,
            height,
            f" {height:.2f}",
            ha="center",
            va="bottom",
            rotation=90,
            fontsize=number_on_top_bar_size
        )

    # Label rects3 (shift slightly right)
    for bar in rects3:
        height = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2 + bar.get_width() * 0.1,
            height,
            f" {height:.2f}",
            ha="center",
            va="bottom",
            rotation=90,
            fontsize=number_on_top_bar_size
        )

    # Axis labels & ticks
    ax.set_ylabel(y_label, fontsize=axis_label_size)
    ax.set_xticks(x)
    ax.set_xticklabels(xtics)

    #ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f"{val:.1f}%"))
    ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f"{val:.1f}×"))
    
    ax.tick_params(axis="x", labelsize=11, rotation=14)
    ax.tick_params(axis="y", labelsize=tick_size)

    ax.set_ylim(0, max(max(y1), max(y2), max(y3)) * 1.15)

    ax.legend(loc="upper left", bbox_to_anchor=(0.1, 1), ncol=1, fontsize=legend_size)

    plt.tight_layout()
    plt.savefig(save_to)
```
