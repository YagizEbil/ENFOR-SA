```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MaxNLocator, FuncFormatter, MultipleLocator
import seaborn as sns
import matplotlib.colors as mcolors
import matplotlib.cm as cm
import statistics
```

```python

def plot_clustered_bars(xtics, y1, y2, y1_std=None, y2_std=None, y1_name="", y2_name="", label1="Series 1", label2="Series 2", save_to="."):
    x = np.arange(len(xtics))  # category positions
    width = 0.35               # width of each bar

    axis_label_size=18
    tick_size=16
    legend_size = 16
    number_on_top_bar_size=12

    #color1, color2 = "blue", "#ff7f0e"
    color1, color2 = "#6082E6", "rosybrown"

    fig, ax = plt.subplots(figsize=(8, 4))
    
    ax2 = ax.twinx()

    # Bars with error bars if provided
    rects1 = ax.bar(x - width, y1, width, label=label1, yerr=y1_std, capsize=5, color=color1, edgecolor="black", hatch="//")
    rects2 = ax2.bar(x + width, y2, width, label=label2, yerr=y2_std, capsize=5, color=color2, edgecolor="black", hatch='xx')

    # Labels and title
    ax.set_ylabel(y1_name, fontsize=axis_label_size,  color=color1) # tamaho to titulo do eixo y. nao dos valores
    ax2.set_ylabel(y2_name, fontsize=axis_label_size, color=color2) # tamaho to titulo do eixo y. nao dos valores
  
    ax.set_xticks(x)
    ax.set_xticklabels(xtics)

    #ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f" {val:.0f}×"))
    ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f" {val:.1f}%"))
    ax2.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f" {val}M"))
    
    # Optional: Add values above bars
    #ax.bar_label(rects1, rotation=90, padding=0, fontsize=number_on_top_bar_size, fmt=" %.1f") # recall: padding=6
    #ax2.bar_label(rects2, rotation=90, padding=0, fontsize=number_on_top_bar_size, fmt=" %.1f") # recall: padding=3

    ax.tick_params(axis='x', labelsize=16, rotation=25)
    ax.tick_params(axis='y', labelsize=tick_size)
    ax2.tick_params(axis='y', labelsize=tick_size)

    #ax.set_ylim(1, max(max(y1), max(y2)) * 1.12) # speedup
    #ax.set_ylim(0, max(max(y1), max(y2)) * 1.12) # avf
    #ax.set_ylim(80, max(max(y1), max(y2)) * 1.05) # recall

    #ax.set_ylim(0, 119.9) # recall
    ax.margins(y=1)   # no vertical margin

    plt.tight_layout(pad=1)
    #ax.legend()
    #ax.legend(loc="lower left", ncol=2, fontsize=legend_size, framealpha=1, facecolor="white") # recall. shift the box a little to the right

    labels = [label1, label2]
    lines = [rects1, rects2]
    ax.legend(lines, labels, loc="upper left", fontsize=14)

    """
    ax.legend(
    loc="upper left",           # anchor legend by its left center
    #bbox_to_anchor=(1, 0.5),     # place it just outside the right edge
    framealpha=1,                # fully opaque
    facecolor="white",
    edgecolor="black",
    fontsize=legend_size
    )
    """
    plt.savefig(save_to) #, dpi=300, bbox_inches='tight')
```
