```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.ticker import MaxNLocator, FuncFormatter, MultipleLocator
```

```python
def plot_lines_and_bar(x, y1, y2, y3, lbl1, lbl2, lbl3, saveto=""):
    xtic_size, ytic_size, axis_size, legend_size = 16, 16, 18, 16

    fig, ax1 = plt.subplots(figsize=(8, 4))

    color1, color2 =  "#6082E6", "rosybrown"
    
    # Left y-axis (two lines)
    ax1.plot(x, y1, label=lbl1, color=color1, marker="s", markersize=9)

    if y2:
        ax1.plot(x, y2, label=lbl2, color=color1, marker="*", markersize=14)

    ax1.set_ylim(0, 30)
        
    #ax1.set_xlabel("X", , fontsize=axis_size)
    ax1.set_ylabel("Injection time",  color=color1, fontsize=axis_size)
    ax1.legend(loc="upper left", fontsize=legend_size)

    # Right y-axis (bar plot)
    ax2 = ax1.twinx()
    ax2.bar(x, y3, alpha=1, color=color2, label=lbl3, capsize=5, edgecolor="black")#, hatch='xx')
    ax2.set_ylabel("Parameters", color=color2, fontsize=axis_size)
    
    ax1.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f" {val}h"))
    ax2.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f" {val}M"))

    ax2.legend(
        loc="lower left",        # anchor point inside axes
        bbox_to_anchor=(0, +0.5), # adjust offset if needed
        fontsize=legend_size
    )
    ax1.tick_params(axis='x', labelsize=xtic_size, rotation=25)
    ax1.tick_params(axis='y', labelsize=ytic_size)
    ax2.tick_params(axis='y', labelsize=ytic_size)

    fig.tight_layout()

    #plt.title("Two Lines (Left Y) and One Bar (Right Y)")
    #plt.show()
    ax1.set_zorder(ax2.get_zorder() + 1)
    ax1.patch.set_visible(False)  # Make sure ax1 background doesn't hide bars
    plt.savefig(saveto)
```

```python
#AREA="gemmini-os-dim8"  # Faults in Gemmini
AREA="sw"       
fn = f"../reports/data/models/{AREA}/scalability.csv"
df = pd.read_csv(fn, comment='#', sep='\t')
```

```python
x = df["Model"].tolist()
y1 = df["Injecton time (Sequential)"].tolist()
y2 = df["Injecton time (Proposed)"].tolist()
y3 = df["Parameters"].tolist()
```

```python
plot_lines_and_bar(x, y1, y2, y3, 
                   lbl1="Injection time (Sequential)", 
                   lbl2="Injection time (Proposed)", 
                   lbl3="Parameters",
                   saveto="../reports/plots/date2026/scalability_sw.pdf"
                  )
```
