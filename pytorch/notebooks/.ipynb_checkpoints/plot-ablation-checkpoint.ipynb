```python
import matplotlib.pyplot as plt
import numpy as np
from tabulate import tabulate
from matplotlib.ticker import MaxNLocator, FuncFormatter, MultipleLocator
import seaborn as sns
import matplotlib.colors as mcolors
import pandas as pd
import matplotlib.cm as cm
```

```python
def plot_bars_by_category(data, formatter=None, id_vars="Layer", ylabel="ylabel", values=["Naive", "Ours",  "Sequential"], save_path="."):
    axis_label_size=14
    tick_size=14
    title_size=14

    # Melt the table into long format
    df_long = data.melt(id_vars=id_vars, value_vars=values, var_name="Condition", value_name="Value")
    
    # Plot 
    plt.figure(figsize=(8, 4))
    ax = sns.barplot(data=df_long, x=id_vars, y="Value", hue="Condition", palette="rocket")
    
    #ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f"{val:.0f}×"))
    #ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f"{val:.1f}%"))
    if formatter:
        ax.yaxis.set_major_formatter(formatter)

    # remove x xics
    #plt.xticks([])

    ax.tick_params(axis='x', labelsize=tick_size)#, rotation=20)
    ax.tick_params(axis='y', labelsize=tick_size)

    #ax.set_ylim(0, 50) # recall

    plt.xlabel("Layers", fontsize=axis_label_size)
    #plt.xlabel("", fontsize=1)
    plt.ylabel(ylabel, fontsize=axis_label_size)

    plt.legend(title=None, fontsize=tick_size, ncol=3)
    plt.tight_layout()
    #print(f"Saving plot to {save_path}")
    #plt.savefig(save_path)
```

```python
def plot_clustered_bars3(
    xtics, y1, y2, y3,
    y_label,
    y1_std=None, y2_std=None, y3_std=None,
    label1="Series 1", label2="Series 2", label3="Series 3",
    save_to="."
):
    width = 0.25               # narrower width (3 bars per group)

    axis_label_size=18
    tick_size=16
    legend_size = 16
    number_on_top_bar_size=12

    # choose colors
    color1, color2, color3 = "#0077b3", "#c6c6c6", "#ff9933"

    fig, ax = plt.subplots(figsize=(8, 4))

    # Three bar groups per x-tick
    rects1 = ax.bar(x - width, y1, width, label=label1, yerr=y1_std, capsize=5, color=color1, edgecolor="black", hatch="//")
    rects2 = ax.bar(x,         y2, width, label=label2, yerr=y2_std, capsize=5, color=color2, edgecolor="black", hatch="xx")
    rects3 = ax.bar(x + width, y3, width, label=label3, yerr=y3_std, capsize=5, color=color3, edgecolor="black", hatch="\\\\")  # double backslash for escape

    """
    # Label rects1 (centered)
    ax.bar_label(rects1, rotation=90, padding=0, fontsize=number_on_top_bar_size, fmt=" %.2f")
    
    # Label rects2 (shift slightly left so they don't collide with rects3)
    for bar in rects2:
        height = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2 - bar.get_width() * 0.1,
            height,
            f" {height:.2f}",
            ha="center",
            va="bottom",
            rotation=90,
            fontsize=number_on_top_bar_size
        )

    # Label rects3 (shift slightly right)
    for bar in rects3:
        height = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2 + bar.get_width() * 0.1,
            height,
            f" {height:.2f}",
            ha="center",
            va="bottom",
            rotation=90,
            fontsize=number_on_top_bar_size
        )
    """
    # Axis labels & ticks
    ax.set_ylabel(y_label, fontsize=axis_label_size)
    ax.set_xlabel("Layers", fontsize=axis_label_size)
    ax.set_xticks([])
    ax.set_xticklabels([])

    #ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f"{val:.1f}%"))
    #ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f"{val:.1f}×"))
    
    ax.tick_params(axis="x", labelsize=14, rotation=0)
    ax.tick_params(axis="y", labelsize=tick_size)
    
    ax.set_ylim(0, max(max(y1), max(y2), max(y3)) * 1.15)

    ax.legend(loc="upper right", ncol=3, fontsize=legend_size)

    plt.tight_layout()
    plt.savefig(save_to)
```

```python
model = "ResNet50"
```

```python
input_file = f"./reports/data/{model}/ablation/ablation_time.csv"
plot_path = f"./reports/plots/ablation/injection_time/{model}.pdf"  # cf or injection_time
df = pd.read_csv(input_file, comment='#', sep='\t')

x = np.arange(1,15)
y1 = df["Naive"].tolist()
y2 = df["Ours"].tolist()
y3 = df["Sequential"].tolist()

#plot_bars_by_category(df, ylabel="Injection time (min)", save_path=plot_path)    
plot_clustered_bars3(x, y1, y2, y3,
                    y_label="Injection time (minutes)", 
                    label1="Naive", 
                    label2="Ours", 
                    label3="Sequential", 
                    save_to="reports/plots/date2026/ablation_time.pdf")
```
